```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import cv2
import warnings
warnings.filterwarnings('ignore')

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, precision_recall_fscore_support
from sklearn.utils.class_weight import compute_class_weight

print("TensorFlow:", tf.__version__)
print("Keras:", keras.__version__)

images = []
labels = []

tumor_path = '/kaggle/input/brain-cancer-mri-dataset/Brain_Cancer raw MRI data/Brain_Cancer'

for subfolder in os.listdir(tumor_path):
    subfolder_path = os.path.join(tumor_path, subfolder)
    for image_filename in os.listdir(subfolder_path):
        image_path = os.path.join(subfolder_path, image_filename)
        images.append(image_path)
        labels.append(subfolder)

normal_mri_path = "/kaggle/input/brain-mri-images/GAN-Traning Images"

for image_filename in os.listdir(normal_mri_path):
    image_path = os.path.join(normal_mri_path, image_filename)
    images.append(image_path)
    labels.append("normal")

df = pd.DataFrame({'image': images, 'label': labels})
print("Original Class Distribution:")
print(df['label'].value_counts())
min_count = min(df['label'].value_counts())
print(f"\nMinimum class count: {min_count}")
normal_multiplier = 1.25

balanced_dfs = []
for class_name in df['label'].unique():
    class_df = df[df['label'] == class_name]
    if class_name == "normal":

        sample_size = int(min_count * normal_multiplier)
        if len(class_df) > sample_size:
            class_df = class_df.sample(n=sample_size, random_state=101)
      
    else:
        
        class_df = class_df.sample(n=min_count, random_state=101)
    balanced_dfs.append(class_df)

df_balanced = pd.concat(balanced_dfs)
df_balanced = df_balanced.sample(frac=1, random_state=101).reset_index(drop=True)

print("\nNew Class Distribution:")
print(df_balanced['label'].value_counts())

X_train, X_test1, y_train, y_test1 = train_test_split(
    df_balanced['image'], df_balanced['label'], 
    test_size=0.2, random_state=101, 
    shuffle=True, stratify=df_balanced['label']
)

X_val, X_test, y_val, y_test = train_test_split(
    X_test1, y_test1, 
    test_size=0.5, random_state=101, 
    shuffle=True, stratify=y_test1
)

df_train = pd.DataFrame({'image': X_train, 'label': y_train})
df_val   = pd.DataFrame({'image': X_val, 'label': y_val})
df_test  = pd.DataFrame({'image': X_test, 'label': y_test})

NUM_CLASSES = len(df_balanced['label'].unique())
print(f"\nNumber of classes: {NUM_CLASSES}")

image_size = (224, 224)
batch_size = 32
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,
    brightness_range=[0.8, 1.2],
    fill_mode='nearest'
)

val_test_datagen = ImageDataGenerator(
    rescale=1./255
)

train_generator = train_datagen.flow_from_dataframe(
    df_train, x_col='image', y_col='label',
    target_size=image_size, batch_size=batch_size,
    class_mode='categorical', shuffle=True
)

val_generator = val_test_datagen.flow_from_dataframe(
    df_val, x_col='image', y_col='label',
    target_size=image_size, batch_size=batch_size,
    class_mode='categorical', shuffle=True
)

test_generator = val_test_datagen.flow_from_dataframe(
    df_test, x_col='image', y_col='label',
    target_size=image_size, batch_size=batch_size,
    class_mode='categorical', shuffle=False
)

class_weights = compute_class_weight(
    'balanced',
    classes=np.unique(y_train),
    y=y_train
)
class_weight_dict = dict(enumerate(class_weights))
print(f"\nClass weights: {class_weight_dict}")
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, GlobalAveragePooling2D
from tensorflow.keras.models import Sequential
from tensorflow.keras.regularizers import l2

def create_custom_model(input_shape=(224, 224, 3), num_classes=4):
    model = Sequential([
        Conv2D(32, (3, 3), activation='relu', padding='same', 
               kernel_regularizer=l2(0.001), input_shape=input_shape),
        BatchNormalization(),
        Conv2D(32, (3, 3), activation='relu', padding='same',
               kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        MaxPooling2D(2, 2),
        Dropout(0.25),
        
        Conv2D(64, (3, 3), activation='relu', padding='same',
               kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        Conv2D(64, (3, 3), activation='relu', padding='same',
               kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        MaxPooling2D(2, 2),
        Dropout(0.25),
    
        Conv2D(128, (3, 3), activation='relu', padding='same',
               kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        Conv2D(128, (3, 3), activation='relu', padding='same',
               kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        MaxPooling2D(2, 2),
        Dropout(0.25),
        Conv2D(256, (3, 3), activation='relu', padding='same',
               kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        Conv2D(256, (3, 3), activation='relu', padding='same',
               kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        MaxPooling2D(2, 2),
        Dropout(0.25),
        
        GlobalAveragePooling2D(),
        Dense(512, activation='relu', kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        Dropout(0.5),
        Dense(256, activation='relu', kernel_regularizer=l2(0.001)),
        BatchNormalization(),
        Dropout(0.5),
        Dense(num_classes, activation='softmax')
    ])
    
    return model
model = create_custom_model(input_shape=(image_size[0], image_size[1], 3), 
                           num_classes=NUM_CLASSES)

checkpoint_cb = ModelCheckpoint("model_cnn_4class.keras", 
                               save_best_only=True, 
                               monitor='val_accuracy',
                               mode='max')

early_stopping_cb = EarlyStopping(patience=15, 
                                 restore_best_weights=True, 
                                 monitor='val_loss',
                                 mode='min')

reduce_lr = ReduceLROnPlateau(monitor='val_loss', 
                             factor=0.2, 
                             patience=5, 
                             min_lr=1e-7)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
print("\n=== Starting Model Training ===")
epochs = 25

history = model.fit(
    train_generator,
    epochs=epochs,
    validation_data=val_generator,
    class_weight=class_weight_dict,
    callbacks=[checkpoint_cb, early_stopping_cb, reduce_lr],
    verbose=1
)

plt.figure(figsize=(15, 6))

plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss', linewidth=2)
plt.plot(history.history['val_loss'], label='Validation Loss', linewidth=2)
plt.legend()
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Train Accuracy', linewidth=2)
plt.plot(history.history['val_accuracy'], label='Validation Accuracy', linewidth=2)
plt.legend()
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
print("\n=== Model Evaluation ===")
test_loss, test_acc = model.evaluate(test_generator)
print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")

y_test_true = test_generator.classes
pred_probs = model.predict(test_generator)
y_pred_classes = np.argmax(pred_probs, axis=1)

class_labels = list(train_generator.class_indices.keys())

def calculate_detailed_metrics(y_true, y_pred, labels):
    """
    Calculate detailed metrics including recall, f1, f2, fp, fn, f0.5
    """
    report = classification_report(y_true, y_pred, target_names=labels, output_dict=True)
    
    cm = confusion_matrix(y_true, y_pred)
    
    fp = cm.sum(axis=0) - np.diag(cm)
    fn = cm.sum(axis=1) - np.diag(cm)
    tp = np.diag(cm)
    tn = cm.sum() - (fp + fn + tp)
    
    precision, recall, f1, support = precision_recall_fscore_support(y_true, y_pred, average=None)
    
    f_beta_2 = []  
    f_beta_05 = []  
    
    for p, r in zip(precision, recall):
        if p + r == 0:
            f_beta_2.append(0)
            f_beta_05.append(0)
        else:
            f_beta_2.append((1+2**2) * (p*r) / ( (2**2)*p + r ))
            f_beta_05.append((1+0.5**2) * (p*r) / ( (0.5**2)*p + r ))
    
    metrics_df = pd.DataFrame({
        'Class': labels,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1,
        'F2-Score': f_beta_2,
        'F0.5-Score': f_beta_05,
        'True_Positives': tp,
        'False_Positives': fp,
        'False_Negatives': fn,
        'True_Negatives': tn,
        'Support': support
    })
    
    return metrics_df, cm, report
detailed_metrics, cm, report_dict = calculate_detailed_metrics(y_test_true, y_pred_classes, class_labels)

print("\n" + "="*60)
print("DETAILED METRICS REPORT")
print("="*60)

print("\n📊 Detailed Metrics per Class:")
pd.set_option('display.float_format', '{:.4f}'.format)
print(detailed_metrics)

print("\n📈 Average Metrics:")
avg_metrics = detailed_metrics[['Precision', 'Recall', 'F1-Score', 'F2-Score', 'F0.5-Score']].mean()
print(avg_metrics)

print("\n📋 Classification Report:")
print(classification_report(y_test_true, y_pred_classes, target_names=class_labels))
plt.figure(figsize=(20, 15))

plt.subplot(2, 2, 1)
sns.heatmap(cm, annot=True, fmt='g', cmap='Blues', 
            xticklabels=class_labels, yticklabels=class_labels)
plt.title("Confusion Matrix", fontsize=14, fontweight='bold')
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.subplot(2, 2, 2)
metrics_to_plot = detailed_metrics[['Precision', 'Recall', 'F1-Score', 'F2-Score', 'F0.5-Score']]
x_pos = np.arange(len(class_labels))
width = 0.15

for i, metric in enumerate(metrics_to_plot.columns):
    plt.bar(x_pos + i*width, metrics_to_plot[metric], width, label=metric)

plt.title("Metrics Comparison by Class", fontsize=14, fontweight='bold')
plt.xlabel("Classes")
plt.ylabel("Score")
plt.xticks(x_pos + 2*width, class_labels, rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(2, 2, 3)
df_balanced['label'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Balanced Class Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Classes")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.subplot(2, 2, 4)
x = np.arange(len(class_labels))
plt.bar(x - 0.2, detailed_metrics['Precision'], 0.2, label='Precision', alpha=0.8)
plt.bar(x, detailed_metrics['Recall'], 0.2, label='Recall', alpha=0.8)
plt.bar(x + 0.2, detailed_metrics['F1-Score'], 0.2, label='F1-Score', alpha=0.8)

plt.title("Precision, Recall, F1-Score by Class", fontsize=14, fontweight='bold')
plt.xlabel("Classes")
plt.ylabel("Score")
plt.xticks(x, class_labels, rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
model.save("brain_mri_4class_balanced_model.h5")

detailed_metrics.to_csv("detailed_metrics_report.csv", index=False)

print(f"\n✅ Model saved as: brain_mri_4class_balanced_model.h5")
print(f"📊 Detailed metrics saved as: detailed_metrics_report.csv")
print("\n" + "="*60)
print("FINAL TRAINING SUMMARY")
print("="*60)
print(f"✅ Classes balanced: {len(df_balanced)} total images")
print(f"✅ Training samples: {len(df_train)}")
print(f"✅ Validation samples: {len(df_val)}")
print(f"✅ Test samples: {len(df_test)}")
print(f"✅ Final Test Accuracy: {test_acc:.4f}")
print(f"✅ Best Validation Accuracy: {max(history.history['val_accuracy']):.4f}")
print("="*60)
```
